```python
This project analyzes temperature data from NOAA's Global Historical Climatology Network (GHCN-Daily). 
The dataset includes temperature records near Ann Arbor, Michigan, spanning from 2005 to 2015. The objective is to visualize:

1)Record High and Low Temperatures (2005-2014)

2)Temperature Records Broken in 2015

3)Weather Stations Near Ann Arbor, Michigan

4)Temperature Summary for 2015

**DATASET**

The analysis is based on two CSV files:

temperature.csv: Contains daily temperature readings (TMAX, TMIN) from various stations.

BinSize.csv: Provides station metadata, including latitude, longitude, and station ID.

**IMPLEMENTATION DETAILS**

The Jupyter Notebook (temp_analysis.ipynb) follows these steps:

1)Data Preprocessing:

  Converts date formats

  Extracts year and day-of-year for analysis

  Removes leap days (February 29)

2)Temperature Trends (2005-2014) & Record Breaks (2015):

  Computes record highs and lows from 2005-2014

  Identifies record-breaking temperatures in 2015

  Visualizes trends using a shaded line graph with scatter points for broken records

3)Weather Station Mapping:

  Uses geopandas and matplotlib to visualize station locations

  Displays stations near Ann Arbor on a simple scatter plot

4)Temperature Summary for 2015:

  Computes daily temperature extremes (max/min) for 2015

  Plots them to illustrate annual temperature variations

**LIBRARIES USED**

1)pandas: Data manipulation

2)numpy: Numerical computations

3)matplotlib: Visualization

4)geopandas & contextily (optional): Map plotting
```
